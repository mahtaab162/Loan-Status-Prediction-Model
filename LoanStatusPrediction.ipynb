## 2. Remove Outliers

```python
# Caluclatte Quartile
```

```python
R_Q1 = df["residential_assets_value"].quantile(0.25)
R_Q3 = df["residential_assets_value"].quantile(0.75)
```

```python
# Calculate IQR
IQR = R_Q3 - R_Q1
IQR
```

```python
lower_limit = R_Q1 - 1.5 * IQR

upper_limit = R_Q3 + 1.5 * IQR
```

```python
df = df[(df["residential_assets_value"] >= lower_limit) & (df["residential_assets_value"] <= upper_limit)]
df
```

```python
plt.figure(figsize=(10,5))
df.boxplot()
plt.xticks(rotation = 40 , ha = "right")
plt.show()
```

## 5. Model Building

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
# Seperating dependent and independent column
```

```python
X = df.drop("loan_status" , axis = 1)
y = df["loan_status"]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X, y, test_size = 0.2, random_state = 39)

models = {"Logistic Regression" : LogisticRegression(max_iter= 1000) ,
          "Decision Tree" : DecisionTreeClassifier(),
          "Random Forest" : RandomForestClassifier()
          }

for name, models in models.items():
  models.fit(X_train , y_train)
  preds = models.predict(X_test)
  print(name , "Accuracy" , accuracy_score (y_test , preds))
```

```python
# Ramdom Forest algorithm wins as it has hoghest accuracy score
```

# Step 1: Features and Targets

```python
X = df.drop("loan_status" , axis = 1)
y = df["loan_status"]
```

# Step 2: Train Test Split

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X, y, test_size = 0.2, random_state = 39 , stratify = y)
```

# Step 3: Build Random Forest Model

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(n_estimators= 200, max_depth= 10, random_state = 39)
```

```python
rf_model.fit(X_train, y_train)
```

# Step 4: Make Predictions

```python
y_pred = rf_model.predict(X_test)
```

# Step 5 : Evaluate the model

```python
from sklearn.metrics import accuracy_score , classification_report , confusion_matrix
print("Accuracy : " , accuracy_score(y_test , y_pred) * 100 , "%" )
print("Classification Report : ", classification_report(y_test , y_pred))
print("Confusion Matrix : " , confusion_matrix(y_test , y_pred))
```
